# Análisis Exploratorio de Datos 

## Importando bibliotecas 


```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

parent_directory=os.path.dirname(os.getcwd())
path=os.path.join(parent_directory,'src')
sys.path.append(parent_directory)
import src.cleaning_data as cln
```

## Conociendo los datos 

```python
print("Cargando el archivo CSV")
BASE_DIR = os.path.dirname(os.getcwd())
proccessed_file=os.path.join(BASE_DIR,"data","processed","processed.csv")

raw_df = pd.read_csv(proccessed_file).copy()
processed_ventas_df = cln.cleaning_data_frame_by_category(raw_df,"venta",min_outliner=1000,max_outliner=1000000)
processed_alquiler_df = cln.cleaning_data_frame_by_category(raw_df,"alquiler",min_outliner=10,max_outliner=1000)

df = processed_ventas_df
print("Mostrando primeras filas")
display(df.head())

print("Mostrando el número de  filas y columnas")
print(f"Filas {df.shape[0]} y columnas {df.shape[1]}")
print("Mostrando el tipos de datos")
print(df.dtypes) 

for i in range(10):
    elemnto = df['Ubicacion'].iloc[i]
    print(elemnto[0])
```

### Medidas de tendencia central y dispersion ventas 

```python
df_venta=processed_ventas_df["Precio"]
print(df_venta.shape[0])

print("Mostrando el promedio")
print(df_venta.mean())
print()
print("Mostrando la mediana")
print(df_venta.median())
print()

print("Mostrando la moda ")
print(df_venta.mode())
print("Mostrando el máximo de las variables")
print(df_venta.max())
print()

print("Mostrando el mínimo de las variables")
print(df_venta.min())
print()

print()

print("Mostrando la desviasion estandar")
print(df_venta.std())

print()
print("Rango intercualtilico precios de casas en venta")
print("Rango intercualtilico 0.75")
print(df_venta.quantile(0.75))
print("Rango intercualtilico 0.25")
print(df_venta.quantile(0.25))

print("Rango intercualtilico rango")
print(df_venta.quantile(0.75)-df_venta.quantile(0.25))

print("Vizualizacion de los precios de las ventas")

plt.figure(figsize=(12, 6))
plt.title("Distribución de Precios de Propiedades en Venta", fontsize=14)
plt.xlabel("Precio (USD)", fontsize=12)
plt.ylabel("Frecuencia", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.3)
sns.histplot(data=df_venta,bins=100)
plt.axvline(df_venta.mean(), color="red", linestyle="--", label=f'Media: {df_venta.mean():.2f}')
plt.axvline(df_venta.median(), color="green", linestyle="-.", label=f'Mediana: {df_venta.median():.2f}')
plt.axvline(df_venta.quantile(0.25), color="purple", linestyle=":", label="Q25")
plt.axvline(df_venta.quantile(0.75), color="orange", linestyle=":", label="Q75")
plt.legend()
plt.show()
```
